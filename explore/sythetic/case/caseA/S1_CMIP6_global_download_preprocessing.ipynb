# S1 — CMIP6 global download and preprocessing / CMIP6全球数据下载与预处理

This notebook downloads and preprocesses CMIP6 data for all selected model families. It reads the download manifest produced by S0.1, queries ESGF for file-level records, downloads missing files with checksum verification, builds annual and climatological products, attaches cell area / land fraction / land-ice fraction, and performs basic QC.

**Inputs:** `S0.1_download_available.csv` and `S0.1_selected_models.csv` from the S0.1 notebook.

**Scope boundary:** global data only. This notebook does not define climate regions, classify land ice, sample grid cells, construct variable pairs, or call `classifier.py`. Negative `evspsbl` values retain their original CMIP6 sign.

**Batch mode:** 35 families are available; `BATCH_FAMILIES` controls which families to process in each run. Start with ~11 representative families, then expand.

**中文说明：** 本Notebook读取S0.1产出的下载清单，从ESGF批量下载和预处理CMIP6数据。支持分批处理——`BATCH_FAMILIES`控制每次运行处理哪些家族，先从11个代表性家族开始，后续扩展到全部35个。下载目标保存到外部硬盘`/Volumes/mimi-T9/CMIP6`。每个模型下载3个core变量（pr/evspsbl/mrro）、最多15个context变量和2~3个固定场，P+R模型跳过evspsbl。18个时间变量覆盖水收支核心（P/ET/R）、能量收支驱动（辐射+湍流通量）、状态控制（温度/土壤水/LAI/雪水当量）和子过程分解（地表径流/降雪/土壤蒸发/蒸腾）。

## 2. Discover file-level records from ESGF / 查询ESGF文件级记录

For each model, the member and grid are already selected by S0.1. This step queries ESGF for the actual file-level records (URLs, checksums, sizes) needed to download each available variable. Fixed fields (areacella, sftlf, sftgif) are **always** queried from ESGF regardless of the S0.1 manifest, because S0.1 checks at the Dataset level while file-level availability may differ, and fx files may live under a different experiment.

**中文说明：** S0.1已经为每个模型选好了member和grid。时变变量按S0.1清单中标记为`available`的逐一查询ESGF；固定场（areacella、sftlf、sftgif）则**始终直接查询ESGF**，因为S0.1是Dataset级检查，文件级可用性可能不同，且fx常在别的experiment下发布。单个模型发现失败不会中断整批。

```python
import gc

def _merge_and_save(new_df: pd.DataFrame, path: Path, key_cols: list[str]) -> pd.DataFrame:
    """Merge new rows into an existing CSV, replacing rows with matching keys."""
    if path.exists():
        existing = pd.read_csv(path)
        existing = existing[~existing.set_index(key_cols).index.isin(new_df.set_index(key_cols).index)]
        merged = pd.concat([existing, new_df], ignore_index=True)
    else:
        merged = new_df
    merged.to_csv(path, index=False)
    return merged


RUNS = []
plan_rows = []
discovery_t0 = time.time()

for model in MODELS:
    info = MODEL_INFO.loc[model]
    member_id = info['member_id']
    pool = info['pool']
    model_manifest = manifest_all[manifest_all['model'] == model]
    available_vars = set(
        model_manifest.loc[model_manifest['status'] == 'available', 'variable'].values
    )
    allow_esgf_without_manifest = (
        model in EXTRA_MODEL_NAMES or model_manifest.empty
    )

    print(f'\n{"="*60}')
    print(f'Resolving {model} (family={info["family"]}, pool={pool}, member={member_id})')
    if allow_esgf_without_manifest and model_manifest.empty:
        print('  no S0.1 manifest rows; will query ESGF for each target variable')

    model_time_vars = {}
    for alias, spec in TIME_VARIABLES.items():
        var_id = spec['variable_id']
        if var_id == 'evspsbl' and pool == 'P+R':
            print(f'  {alias} ({var_id}): skipped (P+R model, no evspsbl)')
            continue
        if var_id not in available_vars and not allow_esgf_without_manifest:
            print(f'  {alias} ({var_id}): not available per S0.1 manifest')
            continue
        model_time_vars[alias] = spec

    records_by_alias = {}
    discovery_errors = {}
    grid_label = None

    for alias, spec in model_time_vars.items():
        var_id = spec['variable_id']
        table_id = spec['table_id']
        var_rows = model_manifest[
            (model_manifest['variable'] == var_id) &
            (model_manifest['table_id'] == table_id) &
            (model_manifest['status'] == 'available')
        ]
        if var_rows.empty:
            if not allow_esgf_without_manifest:
                print(f'  {alias} ({var_id}): no manifest row, skipping')
                continue
            gl = grid_label
        else:
            gl = str(var_rows.iloc[0]['grid_label']).split(',')[0].strip()
            if grid_label is None:
                grid_label = gl

        try:
            records = discover_file_records(
                source_id=model,
                experiment_id=EXPERIMENT_ID,
                variable_id=var_id,
                allowed_tables=[table_id],
                requested_start_year=START_YEAR,
                requested_end_year=END_YEAR,
                fixed=False,
                member_id=member_id,
                activity_id=ACTIVITY_ID,
            )
            if gl:
                records = [
                    r for r in records
                    if r.member_id == member_id and r.grid_label == gl
                ]
            else:
                records = [r for r in records if r.member_id == member_id]
                if records and grid_label is None:
                    grid_label = records[0].grid_label
                    gl = grid_label
                    records = [r for r in records if r.grid_label == gl]
        except Exception as exc:
            records = []
            discovery_errors[alias] = str(exc)
        records_by_alias[alias] = records
        print(f'  {alias} ({var_id}/{table_id}): {len(records)} files on {gl}')

    # One fx variable at a time. Never batch: File-level ESGF results explode.
    fixed_records = {}
    for alias, spec in FIXED_FIELDS.items():
        var_id = spec['variable_id']
        table_id = spec['table_id']
        selected = None
        try:
            candidates = discover_file_records(
                source_id=model,
                experiment_id=EXPERIMENT_ID,
                variable_id=var_id,
                allowed_tables=[table_id],
                fixed=True,
                member_id=member_id,
                activity_id=ACTIVITY_ID,
            )
            if not candidates:
                candidates = discover_file_records(
                    source_id=model,
                    experiment_id=None,
                    variable_id=var_id,
                    allowed_tables=[table_id],
                    fixed=True,
                    activity_id='CMIP',
                )
            selected = select_fixed_record(
                candidates,
                selected_grid=grid_label,
                preferred_member=member_id,
            )
        except Exception as exc:
            selected = None
            discovery_errors[alias] = str(exc)
        fixed_records[alias] = selected
        print(f'  {alias} ({var_id}): {"found" if selected else "MISSING"}')

    missing_required = []
    if 'P' not in records_by_alias or not records_by_alias['P']:
        missing_required.append('P')
    if 'R' not in records_by_alias or not records_by_alias['R']:
        missing_required.append('R')
    if pool == 'core' and ('ET' not in records_by_alias or not records_by_alias['ET']):
        missing_required.append('ET')
    for alias, spec in FIXED_FIELDS.items():
        if spec.get('required', False) and fixed_records.get(alias) is None:
            missing_required.append(spec['variable_id'])

    if missing_required:
        status = 'missing_required'
        print(f'  *** SKIPPED: missing required fields: {missing_required}')
    else:
        status = 'ready'

    optional_missing = [
        alias for alias in model_time_vars
        if alias not in records_by_alias or not records_by_alias[alias]
    ]
    for alias in TIME_VARIABLES:
        if alias not in model_time_vars and alias not in optional_missing:
            optional_missing.append(alias)

    n_files = sum(len(recs) for recs in records_by_alias.values())
    n_files += sum(1 for r in fixed_records.values() if r is not None)
    total_bytes = sum(r.size for recs in records_by_alias.values() for r in recs)
    total_bytes += sum(r.size for r in fixed_records.values() if r is not None)

    run_root = DATA_ROOT / model / EXPERIMENT_ID / member_id / (grid_label or 'gn')

    plan_rows.append({
        'model': model,
        'family': info['family'],
        'pool': pool,
        'member': member_id,
        'grid': grid_label,
        'n_time_vars': len(records_by_alias),
        'n_files': n_files,
        'download_size': human_size(total_bytes),
        'optional_missing': ', '.join(optional_missing),
        'fixed_missing': ', '.join(a for a, r in fixed_records.items() if r is None),
        'status': status,
        'errors': '; '.join(f'{k}: {v}' for k, v in discovery_errors.items()),
    })

    if status == 'ready':
        RUNS.append({
            'model': model,
            'family': info['family'],
            'pool': pool,
            'member_id': member_id,
            'grid_label': grid_label,
            'records_by_alias': records_by_alias,
            'fixed_records': fixed_records,
            'run_root': run_root,
            'model_time_vars': model_time_vars,
        })
    gc.collect()
    if DISCOVERY_MODEL_DELAY_SECONDS > 0:
        print(f'  discovery cooldown: {DISCOVERY_MODEL_DELAY_SECONDS}s')
        time.sleep(DISCOVERY_MODEL_DELAY_SECONDS)

discovery_elapsed = time.time() - discovery_t0
print(f'\nDiscovery complete in {discovery_elapsed:.0f}s ({discovery_elapsed/60:.1f} min)')

PLAN = pd.DataFrame(plan_rows)
print(PLAN.to_string(index=False))
if not PLAN.empty:
    PLAN = _merge_and_save(PLAN, DATA_ROOT / 'S1_run_plan.csv', ['model'])
    print(f'{len(RUNS)} models ready for download out of {len(MODELS)} requested.')
    print(f'Run plan: {PLAN["model"].nunique()} models total (merged with previous runs)')
else:
    print('No runnable models found.')
print('Discovery cell finished. Run the download cell next — do not Run All.')

del plan_rows
gc.collect()
```

## 3. Download files / 下载文件

A cached file is reused only after file-size/checksum, NetCDF structure, variable, and metadata checks. New downloads probe the first three mirrors (2 MB range request, timings cached by host), then transfer from the fastest one. Checksum verification still runs twice on a fresh download (`.part`, then the renamed file). `S1_download_status.csv` is updated during the run so file-level progress survives a kernel interruption. Progress prints the host and MB/s for every file.

**中文说明：** 本步骤逐个模型下载。已有文件经过checksum和NetCDF结构验证后直接复用。新文件先对前3个镜像各测2 MB，选最快的再下载。下载期间持续更新`S1_download_status.csv`，即使kernel中断也可查看文件级进度。新鲜下载仍校验两遍（`.part`和改名后的文件）。每个文件会打印来源主机和MB/s。


```python
import gc
from concurrent.futures import ThreadPoolExecutor, as_completed


DOWNLOAD_STATUS_PATH = DATA_ROOT / 'S1_download_status.csv'
DOWNLOAD_STATUS_COLUMNS = [
    'model', 'family', 'member_id', 'grid_label', 'kind', 'alias',
    'variable_id', 'table_id', 'filename', 'status',
    'expected_size_bytes', 'local_exists', 'local_size_bytes',
    'local_path', 'warnings', 'snapshot_at_utc',
]
STATUS_RUNS = list(RUNS)
_download_status_last_write = 0.0


def _download_status_rows():
    snapshot_at = pd.Timestamp.now(tz='UTC').isoformat()
    rows = []
    for status_run in STATUS_RUNS:
        common = {
            'model': status_run['model'],
            'family': status_run['family'],
            'member_id': status_run['member_id'],
            'grid_label': status_run['grid_label'],
        }
        raw_root = status_run['run_root'] / 'raw'

        for alias, spec in TIME_VARIABLES.items():
            records = status_run['records_by_alias'].get(alias, [])
            if not records:
                if alias == 'ET' and status_run['pool'] == 'P+R':
                    status = 'not_applicable'
                elif alias not in status_run['model_time_vars']:
                    status = 'unavailable_manifest'
                else:
                    status = 'not_found'
                rows.append({
                    **common, 'kind': 'time', 'alias': alias,
                    'variable_id': spec['variable_id'],
                    'table_id': spec['table_id'], 'filename': '',
                    'status': status, 'expected_size_bytes': 0,
                    'local_exists': False, 'local_size_bytes': 0,
                    'local_path': '', 'warnings': '',
                    'snapshot_at_utc': snapshot_at,
                })
                continue
            for record in records:
                target = Path(record.local_path) if record.local_path else raw_root / record.variable_id / record.title
                exists = target.is_file()
                status = record.status
                if status == 'planned' and exists:
                    status = 'present_unverified'
                rows.append({
                    **common, 'kind': 'time', 'alias': alias,
                    'variable_id': record.variable_id, 'table_id': record.table_id,
                    'filename': record.title, 'status': status,
                    'expected_size_bytes': int(record.size or 0),
                    'local_exists': exists,
                    'local_size_bytes': target.stat().st_size if exists else 0,
                    'local_path': str(target),
                    'warnings': ' | '.join(record.warnings),
                    'snapshot_at_utc': snapshot_at,
                })

        for alias, spec in FIXED_FIELDS.items():
            record = status_run['fixed_records'].get(alias)
            if record is None:
                rows.append({
                    **common, 'kind': 'fixed', 'alias': alias,
                    'variable_id': spec['variable_id'],
                    'table_id': spec['table_id'], 'filename': '',
                    'status': 'not_found', 'expected_size_bytes': 0,
                    'local_exists': False, 'local_size_bytes': 0,
                    'local_path': '', 'warnings': '',
                    'snapshot_at_utc': snapshot_at,
                })
                continue
            target = Path(record.local_path) if record.local_path else raw_root / record.variable_id / record.title
            exists = target.is_file()
            status = record.status
            if status == 'planned' and exists:
                status = 'present_unverified'
            rows.append({
                **common, 'kind': 'fixed', 'alias': alias,
                'variable_id': record.variable_id, 'table_id': record.table_id,
                'filename': record.title, 'status': status,
                'expected_size_bytes': int(record.size or 0),
                'local_exists': exists,
                'local_size_bytes': target.stat().st_size if exists else 0,
                'local_path': str(target),
                'warnings': ' | '.join(record.warnings),
                'snapshot_at_utc': snapshot_at,
            })
    return rows


def write_download_status(*, force=False):
    global _download_status_last_write
    now = time.monotonic()
    if not force and now - _download_status_last_write < 1.0:
        return
    frame = pd.DataFrame(_download_status_rows(), columns=DOWNLOAD_STATUS_COLUMNS)
    frame = frame.sort_values(['model', 'kind', 'alias', 'filename']).reset_index(drop=True)
    temporary = DOWNLOAD_STATUS_PATH.with_suffix('.csv.tmp')
    frame.to_csv(temporary, index=False)
    temporary.replace(DOWNLOAD_STATUS_PATH)
    _download_status_last_write = now


REQUIRED_TIME = {'P', 'R'}
REQUIRED_FIXED = {'cell_area', 'land_fraction'}


def obtain_local_file(record, target_dir: Path) -> Path:
    target = target_dir / record.title
    if DOWNLOAD_IF_MISSING:
        return download_record(record, target_dir)
    validation = validate_local_netcdf(target, record)
    if not validation.ok:
        raise FileNotFoundError(
            f'Download disabled and cached file is not valid: {target}; '
            + '; '.join(validation.messages)
        )
    record.local_path = str(target)
    record.status = 'existing-verified'
    record.warnings.extend(validation.warnings)
    return target


def _download_one(item):
    kind, alias, record, target_dir = item
    path = obtain_local_file(record, target_dir)
    return kind, alias, record.title, path


def _is_required(kind: str, alias: str, pool: str) -> bool:
    if kind == 'time':
        if alias in REQUIRED_TIME:
            return True
        if alias == 'ET' and pool == 'core':
            return True
        return False
    return alias in REQUIRED_FIXED


write_download_status(force=True)
print(f'Download status: {DOWNLOAD_STATUS_PATH}')

download_t0 = time.time()
FAILED_DOWNLOADS = []

for i, run in enumerate(RUNS):
    model = run['model']
    pool = run['pool']
    raw_root = run['run_root'] / 'raw'
    raw_root.mkdir(parents=True, exist_ok=True)
    local_paths = {}

    print(f'\n[{i+1}/{len(RUNS)}] Downloading {model} → {run["run_root"]}')
    print(f'  workers={DOWNLOAD_WORKERS} (existing files checked first; optional vars may be skipped)')
    model_t0 = time.time()

    try:
        jobs = []
        for alias, records in run['records_by_alias'].items():
            var_id = run['model_time_vars'][alias]['variable_id']
            for record in records:
                jobs.append(('time', alias, record, raw_root / var_id))
        for alias, record in run['fixed_records'].items():
            if record is None:
                local_paths[alias] = None
                continue
            jobs.append(('fixed', alias, record, raw_root / record.variable_id))

        results = {}
        pending = []
        for kind, alias, record, target_dir in jobs:
            target = target_dir / record.title
            if target.exists():
                path = obtain_local_file(record, target_dir)
                results[(kind, alias, record.title)] = path
            else:
                pending.append((kind, alias, record, target_dir))

        required_errors = []
        optional_errors = []
        if pending:
            with ThreadPoolExecutor(max_workers=DOWNLOAD_WORKERS) as pool_ex:
                futures = {pool_ex.submit(_download_one, job): job for job in pending}
                for future in as_completed(futures):
                    kind, alias, record, _target_dir = futures[future]
                    try:
                        _kind, _alias, title, path = future.result()
                        results[(kind, alias, title)] = path
                    except Exception as exc:
                        short = f'{alias}/{record.variable_id}: {exc}'
                        if _is_required(kind, alias, pool):
                            required_errors.append(short)
                        else:
                            optional_errors.append(short)
                            print(f'  skip optional {short}')
                    finally:
                        write_download_status()
            del futures

        if required_errors:
            raise RuntimeError(' ; '.join(required_errors[:8]))

        for alias, records in run['records_by_alias'].items():
            alias_paths = [
                results[('time', alias, record.title)]
                for record in records
                if ('time', alias, record.title) in results
            ]
            if not alias_paths and alias in run['records_by_alias']:
                if _is_required('time', alias, pool):
                    raise RuntimeError(f'required {alias} has no files')
                print(f'  {alias}: skipped (download failed)')
                continue
            local_paths[alias] = alias_paths
            print(
                f'  {alias}: {len(alias_paths)}/{len(records)} files '
                f'({records[0].status if records else "?"})'
            )

        for alias, record in run['fixed_records'].items():
            if record is None:
                continue
            key = ('fixed', alias, record.title)
            if key not in results:
                if _is_required('fixed', alias, pool):
                    raise RuntimeError(f'required {alias} missing')
                local_paths[alias] = None
                print(f'  {alias}: skipped (download failed)')
                continue
            local_paths[alias] = results[key]
            print(f'  {alias}: {record.status}')

        run['local_paths'] = local_paths

        audit_rows = []
        for alias, records in run['records_by_alias'].items():
            for record in records:
                audit_rows.append({
                    'alias': alias, 'variable_id': record.variable_id,
                    'table_id': record.table_id, 'title': record.title,
                    'status': record.status, 'local_path': record.local_path,
                    'checksum_available': bool(record.checksum and record.checksum_type),
                    'warnings': ' | '.join(record.warnings),
                })
        for alias, record in run['fixed_records'].items():
            if record is not None:
                audit_rows.append({
                    'alias': alias, 'variable_id': record.variable_id,
                    'table_id': record.table_id, 'title': record.title,
                    'status': record.status, 'local_path': record.local_path,
                    'checksum_available': bool(record.checksum and record.checksum_type),
                    'warnings': ' | '.join(record.warnings),
                })
        qc_dir = run['run_root'] / 'qc'
        qc_dir.mkdir(parents=True, exist_ok=True)
        pd.DataFrame(audit_rows).to_csv(qc_dir / 'raw_file_audit.csv', index=False)

        print(f'  Done in {time.time() - model_t0:.0f}s')
        if optional_errors:
            print(f'  optional skipped: {len(optional_errors)}')

    except Exception as exc:
        print(f'  *** FAILED after {time.time() - model_t0:.0f}s: {exc}')
        FAILED_DOWNLOADS.append((model, str(exc)[:300]))
        run['local_paths'] = None

    write_download_status(force=True)
    del results, pending, jobs
    gc.collect()
    if DOWNLOAD_MODEL_DELAY_SECONDS > 0 and i < len(RUNS) - 1:
        print(f'  download cooldown: {DOWNLOAD_MODEL_DELAY_SECONDS}s')
        time.sleep(DOWNLOAD_MODEL_DELAY_SECONDS)

RUNS = [r for r in RUNS if r.get('local_paths') is not None]

print(f'\nDownloads complete in {time.time() - download_t0:.0f}s')
print(f'{len(RUNS)} succeeded, {len(FAILED_DOWNLOADS)} failed')
for model, err in FAILED_DOWNLOADS:
    print(f'  FAILED: {model}: {err}')
write_download_status(force=True)
```

## 5. QC summaries / 质量检查汇总

QC is descriptive only. It reports completeness, value ranges, and negative ET frequencies for the core water-cycle variables. Context variables get basic range checks. Maps are drawn for the core variables only (P, ET, R, WB) to keep output manageable.
