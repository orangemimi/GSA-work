## Step 1: Data Preprocessing

### 1.1 删掉30年平均 P=0的情况


```python
import pandas as pd
from pathlib import Path

# 路径
output_dir = Path("./data/preprocessed")
no0_dir = output_dir / "no0"
no0_dir.mkdir(parents=True, exist_ok=True)

# 读取数据
classic_mean = pd.read_csv(output_dir / "classic_preprocessed_30yr_mean.csv")
lpj_mean = pd.read_csv(output_dir / "lpj_guess_preprocessed_30yr_mean.csv")

# 删除 P=0 的样本
classic_mean_no0 = classic_mean[classic_mean["precipitation"] != 0].copy()
lpj_mean_no0 = lpj_mean[lpj_mean["precipitation"] != 0].copy()

# 保存
classic_no0_path = no0_dir / "classic_preprocessed_30yr_mean_no0.csv"
lpj_no0_path = no0_dir / "lpj_guess_preprocessed_30yr_mean_no0.csv"

classic_mean_no0.to_csv(classic_no0_path, index=False)
lpj_mean_no0.to_csv(lpj_no0_path, index=False)

# 打印结果
print("✓ 已保存去除 P=0 的文件")
print(f"CLASSIC: {classic_no0_path}")
print(f"  原始样本数: {len(classic_mean):,}")
print(f"  删除后样本数: {len(classic_mean_no0):,}")
print(f"  删除数量: {len(classic_mean) - len(classic_mean_no0):,}")

print(f"\nLPJ-GUESS: {lpj_no0_path}")
print(f"  原始样本数: {len(lpj_mean):,}")
print(f"  删除后样本数: {len(lpj_mean_no0):,}")
print(f"  删除数量: {len(lpj_mean) - len(lpj_mean_no0):,}")
```
